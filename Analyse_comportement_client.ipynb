## 3. Analyse des relations(Statistique)

## 3.1 Les indicateurs du comportement 

#### L'ANOVA pose essentiellement :

#### H₀ : les moyennes de PA des 5 groupes d'âge sont égales.
#### H₁ : au moins une moyenne est différente.
#### p-value < 0.05 → on rejette H₀

```python
from scipy.stats import f_oneway
```

```python
groupes = df.groupby("Age_Group", observed=True)["Purchase Amount (USD)"]
```

```python
groupes
```

```python
groupes_age = [groupe for _, groupe in groupes]
groupes_age
#une liste de Series Pandas ( inedex et valeur)
```

```python
resultat = f_oneway(*groupes_age)

print(resultat)
```

#### On n’a pas détecté de différence significative du PA moyen selon les groupes d’âge.

```python
from scipy.stats import f_oneway, ttest_ind

facteurs = [
    "Age_Group",
    "Gender",
    "Location",
    "Subscription Status",
    "Item Purchased",
    "Category",
    "Size",
    "Color",
    "Season",
    "Shipping Type",
    "Discount Applied",
    "Payment Method"
]

cibles = ["Purchase Amount (USD)", "Purchase_Frequency_Num"]

for facteur in facteurs:
    for cible in cibles:

        groupes = [
            groupe[cible].dropna()
            for _, groupe in df.groupby(facteur, observed=True)
        ]

        if len(groupes) == 2:
            stat, p = ttest_ind(
                groupes[0],
                groupes[1],
                equal_var=False
            )
            test = "Test t"

        else:
            stat, p = f_oneway(*groupes)
            test = "ANOVA"

        print(
            facteur,
            "→",
            cible,
            "|",
            test,
            "| p-value =",
            round(p, 4)
        )
```

#### Seule la saison est significativement associée au PA ; aucun facteur testé n’est significativement associé à la FA.
